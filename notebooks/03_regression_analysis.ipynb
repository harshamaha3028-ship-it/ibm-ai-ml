**Automobile Efficiency Regressor & Vehicle Class Profiler**
Comparing Linear, Ridge, Lasso, and ElasticNet Regression with 5-Fold Cross Validation.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json

with open('../models/model_metadata.json', 'r') as f:
    meta = json.load(f)

print(f"Selected Best Model: {meta['best_regression_model']}")
pd.DataFrame(meta['model_comparison']).T
```

```python
# Feature coefficients
impacts_df = pd.DataFrame(meta['feature_impacts'])
print(impacts_df)

plt.figure(figsize=(9, 5))
colors = ['#16a34a' if c > 0 else '#dc2626' for c in impacts_df['coefficient']]
plt.barh(impacts_df['feature'], impacts_df['coefficient'], color=colors)
plt.axvline(0, color='grey', linestyle='--')
plt.title('Standardized Coefficients (Feature Impact on MPG)', fontweight='bold')
plt.xlabel('Coefficient Value')
plt.gca().invert_yaxis()
plt.show()
```
