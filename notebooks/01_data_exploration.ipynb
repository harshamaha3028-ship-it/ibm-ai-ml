# 01. Exploratory Data Analysis (Auto MPG Dataset)
**Automobile Efficiency Regressor & Vehicle Class Profiler**
This notebook explores distributions, correlations, and handles missing horsepower values.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Auto MPG dataset
cols = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model_year', 'origin', 'car_name']
df = pd.read_csv('../data/raw/auto-mpg.csv')
print(f'Dataset Shape: {df.shape}')
df.head()
```

```python
# Data types and missing values summary
print(df.info())
print('\nMissing values in horsepower: ', df['horsepower'].isna().sum() if df['horsepower'].dtype != 'O' else (df['horsepower'] == '?').sum())
```

```python
# Convert horsepower to numeric and median impute
df['horsepower'] = pd.to_numeric(df['horsepower'], errors='coerce')
median_hp = df['horsepower'].median()
df['horsepower'] = df['horsepower'].fillna(median_hp)
print(f'Imputed horsepower with median: {median_hp} HP')
df.describe()
```

```python
# Distribution of Target Variable: MPG
plt.figure(figsize=(8, 4))
sns.histplot(df['mpg'], kde=True, color='#2563eb', bins=25)
plt.title('Distribution of Target Variable: Miles Per Gallon (MPG)', fontweight='bold')
plt.xlabel('MPG')
plt.show()
```

```python
# Correlation Matrix Heatmap
plt.figure(figsize=(9, 7))
numeric_df = df.drop(columns=['car_name'])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap of Automobile Characteristics', fontweight='bold')
plt.show()
```
