# 02. Feature Engineering
**Automobile Efficiency Regressor & Vehicle Class Profiler**
Deriving domain-specific automotive features without data leakage.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

sys.path.insert(0, os.path.abspath('..'))
from app.utils.preprocessing import AutoMPGFeatureEngineer

df = pd.read_csv('../data/raw/auto-mpg.csv')
df['horsepower'] = pd.to_numeric(df['horsepower'], errors='coerce')
df['horsepower'] = df['horsepower'].fillna(df['horsepower'].median())

fe = AutoMPGFeatureEngineer(reference_year=82)
df_feat = fe.transform(df)
df_feat['mpg'] = df['mpg']
df_feat.head()
```

```python
# Correlation of engineered features with MPG
corrs = df_feat.corr()['mpg'].sort_values()
print('Correlation of features with MPG:\n', corrs)

plt.figure(figsize=(10, 5))
corrs.drop('mpg').plot(kind='barh', color='#3b82f6')
plt.title('Engineered Features Correlation with MPG', fontweight='bold')
plt.xlabel('Pearson Correlation Coefficient')
plt.show()
```
